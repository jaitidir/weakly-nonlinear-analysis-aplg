# Oscillatory instability (Pe = 6)

```julia
include("../src/oscillatory.jl")

# true: compute that data again and save it in data/oscillatory; false: load the saved data
# (spinodal: seconds; speed and branches: tens of minutes; time stepping: many hours)
COMPUTE = (speed=false, spinodal=false, branches=false, timestepping=false)
```

## Speed of the travelling wave (φ = 0.88)

```julia
# Newton continuation of T in δ (φa = ϕac + δ²), from δ = 0.1 down to 0.01 and up to 0.19
speed = getdata(travelling_speed, "oscillatory/speed_phi0.88.jld2", COMPUTE.speed)
savepdf(speed_figure(speed), "oscillatory_speed_phi0.88")
```

## Spinodals

```julia
# Weakly nonlinear classification of T (a) and CP (b) along the spinodal
classes = getdata(spinodal_classes, "oscillatory/spinodal_classes_Pe6.jld2", COMPUTE.spinodal)
savepdf(spinodals_figure(classes), "oscillatory_spinodals")
```

## Bifurcation diagrams

```julia
# At each φ: Newton continuation of T with the weakly nonlinear T and CP curves, the PS branches, and the time-stepping
# sweep from four initial conditions
sweeps = []
for ϕ in PHIS
    tb = getdata(() -> travelling_branch(ϕ), "oscillatory/travelling_phi$(ϕ).jld2", COMPUTE.branches)
    ps = ps_branches(ϕ, tb, COMPUTE.branches)
    ts = getdata(() -> timestepping_sweep(ϕ, (tb.phia_lo_plot, tb.phia_hi_plot)), "oscillatory/timestepping_phi$(ϕ).jld2",
                 COMPUTE.timestepping)
    push!(sweeps, (ϕ, ts))
    fig, rho = bifurcation_figures(ϕ, tb, ps, ts)
    savepdf(fig, "oscillatory_bifurcation_phi$(ϕ)"); savepdf(rho, "oscillatory_bifurcation_phi$(ϕ)_rho")
    display(fig); display(rho)
end
```

## Shaded spinodal

```julia
savepdf(shaded_spinodal_figure(classes, sweeps), "oscillatory_spinodals_shaded")
```
